## Part 2 — `pandas` for Data Analysis
### Read a CSV


```python
import pandas as pd
df = pd.read_csv('temperatures.csv')
df
```

### Compute stats

```python
df['temp'].mean(), df['temp'].min(), df['temp'].max()
```

### Add column

```python
df['temp_F'] = df['temp']*9/5+32
df
```

### On your own

```python
# Compute the average temperature in Fahrenheit
meanF = df['temp_F'].mean()
print(meanF)
# Save this average temperature to a new file, "average_temp.txt"
with open("average_temp.txt", "w") as file:
    file.write(str(meanF))
```
